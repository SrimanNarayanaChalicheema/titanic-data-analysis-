***Data Cleaning***

```python
print("Missing values count in each column:")
df.isnull().sum()
```

```python
def cleaning(df):
    df = df.copy()
    df["Age"] = df["Age"].fillna(df["Age"].median())
    df["Cabin"] = df["Cabin"].fillna("Unknown")
    df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
    return df
```

```python
train_data = cleaning(df)
train_data
```

For the numerical data, we use median to fill the null values.  
For categorical data like Embarked we use the most embarked port. i.e, mode()

```python
train_data.isnull().sum()
```

All missing values in the dataset have been handled by the cleaning function. The resulting dataset contains no missing values in any column.

```python
plt.hist(train_data["Fare"], bins = "scott",edgecolor = 'black')

plt.title("Distribution of Passenger Fares")
plt.xlabel("Fare")
plt.ylabel("Number of Passengers")

plt.show()
```

Observation: Many passengers paid low fare and some passengers paid higher fare

```python
print("Survival rate of passengers who paid higher fare:")

train_data["Fare_Range"] = pd.cut(
    train_data["Fare"],
    bins=[0, 10, 30, 100,float("inf")],
    labels=["Low", "Medium", "High", "Very High"],
    include_lowest=True,
    right=False
)

survival_rate = train_data.groupby("Fare_Range", observed=False)["Survived"].mean() * 100

print(survival_rate)
plt.bar(survival_rate.index, survival_rate.values)

plt.title("Survival Rate by Fare Range")
plt.xlabel("Fare Range")
plt.ylabel("Survival Rate (%)")
plt.xticks(rotation=55)
plt.show()
```

* observed=True returns only categories present in the data.
* observed=False includes all defined categories, even if some
* categories have no observations.

Observation: Survival rates increased across the defined fare ranges in this dataset. This indicates an association between fare paid and survival, but does not establish that paying a higher fare directly caused a passenger to survive.

## Data-Cleaning Decisions

1. **Missing Age values:** Filled missing ages using the median age because the median is less affected by extreme values.
2. **Missing Embarked values:** Filled missing embarkation ports using the mode, the most frequently occurring category.
3. **Missing Cabin values:**  Replaced missing values with "Unknown" to distinguish unavailable cabin information from recorded cabin values.
4. **Duplicate records:** Checked the dataset for duplicate rows. No duplicate rows were found, so no records needed to be removed.
   
These steps help improve data quality while minimizing unnecessary information loss.
